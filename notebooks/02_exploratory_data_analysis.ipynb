# Exploratory Data Analysis & Market Insights

This notebook explores the processed Indian tech job market dataset to answer core analytical questions:
- Which skills are most demanded across Data Analyst, Data Scientist, and ML Engineer positions?
- How do salaries vary across major tech hubs (Bangalore, Mumbai, Delhi NCR)?
- What is the distribution of compensation between internships and full-time roles?
- Which technical skill pairs most frequently co-occur?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

PROCESSED_DIR = Path('../data/processed')
df_jobs = pd.read_csv(PROCESSED_DIR / 'jobs_cleaned.csv')
df_skills = pd.read_csv(PROCESSED_DIR / 'job_skills.csv')
```

### 1. In-Demand Skills Analysis

```python
top_skills = df_skills['skill'].value_counts().head(15)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_skills.values, y=top_skills.index, color='#1f77b4')
plt.title('Top 15 Most Demanded Skills in Job Postings')
plt.xlabel('Frequency')
plt.ylabel('Skill')
plt.tight_layout()
plt.show()
```

### 2. Salary Distribution by Role Category

```python
sal_df = df_jobs.dropna(subset=['salary_avg'])
plt.figure(figsize=(10, 5))
sns.boxplot(data=sal_df, x='role_category', y='salary_avg', palette='Set2')
plt.title('Salary Distribution by Role (Lakhs Per Annum)')
plt.xlabel('Role Category')
plt.ylabel('Average Salary (LPA)')
plt.tight_layout()
plt.show()
```

### 3. Geographic Hubs & Hiring Volume

```python
top_cities = df_jobs['city'].value_counts().head(8)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_cities.index, y=top_cities.values, color='#2ca02c')
plt.title('Top Hiring Hubs by Listing Volume')
plt.xlabel('Location')
plt.ylabel('Listings Count')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### 4. Skill Co-occurrence Heatmap

```python
top_8_skills = df_skills['skill'].value_counts().head(8).index
merged = pd.merge(df_jobs[['job_id', 'role_category']], df_skills, on='job_id')
filtered = merged[merged['skill'].isin(top_8_skills)]
ct = pd.crosstab(filtered['role_category'], filtered['skill'])

plt.figure(figsize=(9, 5))
sns.heatmap(ct, annot=True, fmt='d', cmap='Blues')
plt.title('Skill Demand Breakdown Across Roles')
plt.tight_layout()
plt.show()
```
